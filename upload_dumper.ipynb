# Notebook that collects data (biased sample) for hand in. 

Not a very interesting notebook; this just collects at least one model of all model entries and copies it to another folder.

```python
import os
import sys
from tqdm import tqdm
import pandas as pd
sys.path.append('utils')
from  configloader import Configloader
from database import Database
import file_io
```

```python
config = Configloader()

db = Database(config)
db.connect()
basedir = config.get('settings', 'image_directory')

```

```python
query = "SELECT brand, model, year, MIN(autoscout_id) as l  FROM listings GROUP BY brand, model, year"
rows = db.execute_query(query)
```

```python
dfall = pd.DataFrame(rows)
```

```python
dfall
```

```python
#still a bit of a big file, let's select a max of 15 years if there are more than 15 years per brand,model combo: 

for _, df in tqdm(dfall.groupby(['brand','model'])):
    if df.year.nunique() > 15: 
        df = df.sample(15)
    for i, row in df.iterrows(): 
        # row = row[1]
        source = os.path.join(basedir, row['brand'], row['model'], row['l'])
        target = os.path.join('/home/frederic/Downloads/SyntraDump/',  row['brand'], row['model'], row['l'])
        for file in os.listdir(source): 
            sourcefile = os.path.join(source, file)
            targetfile = os.path.join(target, file)
            file_io.copy_file(sourcefile, targetfile)
```
